```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = 'https://countrycode.org/'
url
```

```python
page = requests.get(url)
page
```

```python
soup = BeautifulSoup(page.text, 'html.parser')
```

```python
table = soup.find('table')
```

```python
columns_header = table.find_all('th')
```

```python
columns_title = [i.text.strip() for i in columns_header]
```

```python
clean_columns = []
for title in columns_title:
    if title and title not in clean_columns:
        clean_columns.append(title)
```

```python
df = pd.DataFrame(columns=clean_columns)
```

```python
data = table.find_all('tr')
```

```python
for row in data:
    row_data = row.find_all('td')    
    if len(row_data) == len(clean_columns):
        ind_row_data = [x.text.strip() for x in row_data]
        length = len(df)
        df.loc[length] = ind_row_data
```

```python
df.to_csv('Country_Codes.csv', index = False)
```
